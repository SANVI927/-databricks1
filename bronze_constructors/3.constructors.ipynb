```python
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,DoubleType
constructors_schema=StructType(fields=[StructField("constructorId",IntegerType(),True),
                                      StructField("constructorRef",StringType(),True),
                                      StructField("name",StringType(),True),
                                      StructField("nationality",StringType(),True),
                                      StructField("url",StringType(),True)
])
```

```python
constructors_df=spark.read\
    .schema(constructors_schema)\
        .json("/Volumes/practice_catalog/pyspark_practice/csv_files/constructor.json")
```

```python
constructors_df.display()
```

```python
from pyspark.sql.functions import current_timestamp,current_date
constructors_df_final=constructors_df.withColumn("ingestion_timestamp",current_timestamp())\
                                     .withColumn("ingestion_date",current_date())\
                                     .withColumnRenamed("constructorId","constructor_id")\
                                     .withColumnRenamed("constructorRef","constructor_ref")\
                                     .drop("url")
                                     


                                       
```

```python

display(constructors_df_final)
```

```python
%sql
show tables 
```

```python
constructors_df_final.write\
    .format("delta")\
        .mode("overwrite")\
            .saveAsTable("practice_catalog.pyspark_practice.constructors")


```
