```python
lap_times_df=spark.read\
    .option("inferSchema",True)\
        .csv("/Volumes/practice_catalog/pyspark_practice/csv_files/lap_times.csv")
lap_times_df.display()
```

```python
from pyspark.sql.types import StructType,StructField,IntegerType,StringType
lap_times_schema=StructType(fields=[StructField("raceId",IntegerType(),True),
                                      StructField("driverId",IntegerType(),True),
                                      StructField("lap",IntegerType(),True),
                                      StructField("position",IntegerType(),True),
                                      StructField("time",StringType(),True),
                                      StructField("milliseconds",IntegerType(),True)

                                      
])
```

```python
lap_times_df=spark.read\
    .schema(lap_times_schema)\
        .csv("/Volumes/practice_catalog/pyspark_practice/csv_files/lap_times.csv/")
```

```python
lap_times_df.display()
```

```python
lap_times_df.printSchema()
```

```python
from pyspark.sql.functions import lit,current_timestamp,current_date
v=lap_times_df.withColumn("ingestion_timestamp",current_timestamp())\
                .withColumn("ingestion_date",current_date())\
                .withColumnRenamed("raceId","race_Id")\
                .withColumnRenamed("driverId","driver_Id")
```

```python

display(v)
```

```python
v.write\
    .format("delta")\
        .mode("overwrite")\
        .option("overwriteSchema", "true")\
            .saveAsTable("practice_catalog.pyspark_practice.laptimes")
```
