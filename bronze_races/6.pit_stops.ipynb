```python
pit_stops_df=spark.read\
    .option("multiLine",True)\
    .option("inferSchema",True)\
        .json("/Volumes/practice_catalog/pyspark_practice/csv_files/pit_stops.json")


```

```python
pit_stops_df.display()
```

```python
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,DoubleType
pit_stops_schema=StructType(fields=[StructField("raceId",IntegerType(),True),
                                      StructField("driverId",IntegerType(),True),
                                      StructField("stop",IntegerType(),True),
                                      StructField("lap",IntegerType(),True),
                                      StructField("time",StringType(),True),
                                      StructField("duration",StringType(),True),
                                      StructField("milliseconds",IntegerType(),True)
])
```

```python
pit_stops_df=spark.read\
    .option("multiLine",True)\
    .schema(pit_stops_schema)\
        .json("/Volumes/practice_catalog/pyspark_practice/csv_files/pit_stops.json")
```

```python
pit_stops_df.display()
```

```python
pit_stops_df.printSchema()
```

```python
from pyspark.sql.functions import col
pit_stops_df=pit_stops_df.withColumnRenamed("raceId","race_Id")\
                                     .withColumnRenamed("driverId","driver_Id")\
                                     
                                     


                                       
```

```python
pit_stops_df.display()
```

```python
from pyspark.sql.functions import current_timestamp,current_date
pit_stops_df=pit_stops_df.withColumn("ingetion_timestamp",current_timestamp())\
    .withColumn("ingestion_date",current_date())


```

```python
pit_stops_df.display
```

```python
pit_stops_df.write\
    .mode("overwrite")\
    .format("delta")\
    .saveAsTable("practice_catalog.pyspark_practice.pit_stops")
```
