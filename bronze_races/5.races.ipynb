```python
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,DoubleType
races_schema=StructType(fields=[StructField("raceId",IntegerType(),True),
                                      StructField("year",IntegerType(),True),
                                      StructField("round",IntegerType(),True),
                                      StructField("circuitId",IntegerType(),True),
                                      StructField("name",StringType(),True),
                                      StructField("date",StringType(),True),
                                      StructField("time",StringType(),True),
                                      StructField("url",StringType(),True)
                                      
])
```

```python
races_df = spark.read\
    .schema(races_schema)\
    .option("header", True)\
    .csv("/Volumes/practice_catalog/pyspark_practice/csv_files/races.csv")

```

```python
races_df.display()
```

```python
from pyspark.sql.functions import current_timestamp,current_date
races_df=races_df.withColumn("ingestion_timestamp",current_timestamp())\
                                     .withColumn("ingestion_date",current_date())\
                                     .withColumnRenamed("raceId","race_Id")\
                                     .withColumnRenamed("circuitId","circuit_Id")\
                                     .drop("url")
                                    
```

```python
races_df.display()
```

```python
races_df.write\
    .format("delta")\
        .mode("overwrite")\
            .option("mergeschema",True)\
            .saveAsTable("practice_catalog.pyspark_practice.races")
```
