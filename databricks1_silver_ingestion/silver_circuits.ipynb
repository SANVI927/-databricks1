```python
df=spark.table("practice_catalog.pyspark_practice.circuit")
display(df)
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import col
df1=df.filter(col("circuitid")==1)
df1.display()
              
```

```python
df1=df.filter(col("circuitid").isNull())
df1.display()
```

```python
df1=df.filter(col("circuitid").isNotNull())
df1.display()
```

```python
df_filter=df.filter((col("circuitid")==1) & (col("circuitref")=='albert_park'))
df_filter.display()
```

```python
df_filter=df.filter((col("circuitid")==1) |(col("circuitref")=='americas')|(col("country")=='UK'))
df_filter.display()
```

```python
df_filter=df.filter(col("country").isin("Australia","Japan","USA")).display()
```

```python
df=df.groupBy(col("circuitId")).count().filter(col("count")>1)
df.display()
```

```python
from pyspark.sql.functions import col,when,lit,concat
df= df.withColumn("circuitRef",when(col("circuitRef")=="albert_park",concat(lit(" "),col("circuitRef"),lit(" "))).otherwise(col("circuitRef")))
df.display()
```

```python
from pyspark.sql.functions import col,upper,lower,initcap,lit,concat,trim,ltrim,rtrim
df= df.withColumn("circuitRef", initcap(col("circuitRef")))\
           .withColumn("circuitRef_upper",upper(col("circuitRef")))\
                .withColumn("circuitRef_lower", lower(col("circuitRef")))\
                    .withColumn("circuitRef_ltrim",ltrim(col("circuitRef")))\
                      .withColumn("circuitRef_rtrim",rtrim(col("circuitRef")))
                      
df.display()
```

```python
from pyspark.sql.functions import length,split,substring,element_at,col
df= df.withColumn("length",length(col("circuitRef")))\
     .withColumn("circuitRef_split",split(col("circuitRef")," "))\
        .withColumn("circuitRef_split_0",split(col("circuitRef"),"_").getItem(0))\
             .withColumn("country_prefix",substring(col("country"),1,3))
              
df.display()
```

```python
from pyspark.sql.functions import contains
df=df.filter(col("circuitRef").contains("a"))\
    .select("circuitRef").display()
```

```python
from pyspark.sql.functions import col,like,startWith
df=df.filter(col("circuitRef").like("%a%"))\
    .select("circuitRef").display()
```

```python
from pyspark.sql.functions import col
df=df.filter(col("circuitRef").endswith("a"))\
    .select("circuitRef").display()
#df=df.filter(col("circuitRef").startswith("a"))\
    #.select("circuitRef").display()
    
```

```python
from pyspark.sql.functions import col,when
df_circuits=df.withColumn("circuitRef_label",when(col("latitude")<1,"low")\
                          .when((col("latitude")>=10) & (col("latitude")<50),"medium")\
                          .otherwise("high"))
df_circuits.display()
```

```python
df_circuits.write.format("delta").mode("overwrite").option("overwriteSchema","true").saveAsTable("practice_catalog.pyspark_practice.silver_circuit")
```
