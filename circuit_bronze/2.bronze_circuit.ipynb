```python
df = spark.read\
    .option("header", True)\
    .option("inferSchema", "True")\
    .option("mode","permissive")\
    .csv("/Volumes/practice_catalog/pyspark_practice/csv_files/circuit.csv")
df.display()  
```

```python
df.printSchema()
```

```python
print(df.columns)
```

```python
df=df.select(
    "circuitId",
    "circuitRef",
    "location",
    "country",
    "lat",
    "lng",
    "alt",
    "url"
)
display(df)
```

```python
from pyspark.sql.functions import lit
df=df.withColumn("source",lit("formula1"))
display(df)
```

```python
from pyspark.sql.functions import current_timestamp
df=df.withColumn("ingestion_date",current_timestamp())

df=df.withColumnRenamed("lat","latitude")\
       .withColumnRenamed("lng","longitude")\
       .withColumnRenamed("alt","altitude")
display(df)
```

```python
df=df.drop("source")
df=df.drop("url")
display(df)
```

```python
spark.sql("show tables").show()
```

```python
df.write\
    .format("delta")\
    .mode("overwrite")\
    .option("overwriteSchema", "true")\
    .saveAsTable("practice_catalog.pyspark_practice.circuit")
```
